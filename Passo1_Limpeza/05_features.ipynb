```python
import pandas as pd
import numpy as np
```

```python
demandas = pd.read_csv('demandas_filtradas.csv', parse_dates=['data_inicio', 'data_fim'])
print(f"{len(demandas)} demandas | colunas: {list(demandas.columns)}")
```

```python
VAZAMENTO = ['qtd_sessoes', 'data_fim']
IDENTIFICADORES = ['chamado_mascarado']

print("Descartadas por vazamento:", VAZAMENTO)
print("Descartadas por serem identificadores:", IDENTIFICADORES)
```

```python
#Remove o prefixo "Chamado_XXX - " que nós mesmos criamos no mascaramento
texto = demandas['descricao'].str.replace(r'^Chamado_\d+ - ', '', regex=True)

demandas['desc_num_caracteres'] = texto.str.len()
demandas['desc_num_palavras'] = texto.str.split().str.len()

#Flags de palavras-chave
palavras = {
    'kw_relatorio':   'relat',
    'kw_dashboard':   'dashboard',
    'kw_powerbi':     'power ?bi',
    'kw_personaliza': 'personaliza',
    'kw_desenvolve':  'desenvolv',
    'kw_ajuste':      'ajust',
    'kw_correcao':    'corri|corre',
    'kw_divergencia': 'divergen',
}

for nome, padrao in palavras.items():
    demandas[nome] = texto.str.contains(padrao, case=False, regex=True).astype(int)

demandas[list(palavras)].sum().sort_values(ascending=False)
```
